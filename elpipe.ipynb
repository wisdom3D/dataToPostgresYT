```python
count = 0
for batch in file.iter_batches(batch_size = 100000):
    count+=1
    batch_df = batch.to_pandas()
    batch_df.to_sql(name = 'ny_taxi_data', con=engine, if_exists = 'append')
print(f'complete {count} batches.')
```
